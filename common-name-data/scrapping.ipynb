```python
# Step 1: Import the modules.
import requests
import pandas as pd
from bs4 import BeautifulSoup
from io import StringIO

def fetch_common_name(url):
  try:
    # Step 2: Send a request to the page.
    response = requests.get(url)
    response.raise_for_status() # check if the request was succesfull
  except requests.exceptions.RequestException as e:
    print(f"Error fetching the webpage: {e}")
    return None
  
  try:
    # Step 3.1: Parse the HTML content with BeautifulSoup.
    soup = BeautifulSoup(response.content, "html.parser")

    # Step 3.2: Find the specific table (usually the first table under this section).
    table = soup.find_all("table", {"id": "myTable"})

    # Step 3.3: Read the table into a DataFrame using pandas.
    df = pd.read_html(StringIO(str(table)))[0]
  except Exception as e:
    print(f"Error parsing the HTML or reading the table: {e}")
    return None
    
  try:
    # Step 4: Clean up the DataFrame if necessary.
    # Clean up any unwanted rows or columns.
    df = df.dropna(how="all")  # Drop rows with all NaN values.
    return df
  except Exception as e:
    print(f"Error cleaning or processing the DataFrame: {e}")
    return None
    
    
```

```python
common_surname_urls = [
  "https://names.mongabay.com/most_common_surnames.htm",
  "https://names.mongabay.com/most_common_surnames1.htm",
  "https://names.mongabay.com/most_common_surnames2.htm",
  "https://names.mongabay.com/most_common_surnames5.htm",
]

all_names = []
for url in common_surname_urls:
  all_names.append(fetch_common_name(url))

surnames_df = pd.concat(all_names)
print(surnames_df)

# Save the DataFrame to a CSV file.
file_name = "common_surnames"
surnames_df.to_csv(f"{file_name}.csv", index=False)
print(f"\nData saved to '{file_name}.csv'.")
```

```python
male_names_url = "https://names.mongabay.com/male_names_alpha.htm"

male_names_df = fetch_common_name(male_names_url)

if male_names_df is not None:
  print("Successfully extracted the table")
  print(male_names_df)
  # Save the DataFrame to a CSV file.
  file_name = "common_male_names"
  male_names_df.to_csv(f"{file_name}.csv", index=False)
  print(f"\nData saved to '{file_name}.csv'.")
```

```python
female_names_url = "https://names.mongabay.com/female_names_alpha.htm"

female_names_df = fetch_common_name(female_names_url)

if female_names_df is not None:
  print("Successfully extracted the table")
  print(female_names_df)
  # Save the DataFrame to a CSV file.
  file_name = "common_female_names"
  female_names_df.to_csv(f"{file_name}.csv", index=False)
  print(f"\nData saved to '{file_name}.csv'.")
```

```python
import math
from random import randint

surnames = surnames_df["Surname"].to_list()
male_names = male_names_df["Name"].to_list()
female_names = female_names_df["Name"].to_list()
```

```python
def generate_random_name(gender=None):
  surname = surnames[randint(0, len(surnames))]
  if gender == "male":
    return male_names[randint(0, len(male_names))] + " " + surname
  elif gender == "female":
    return female_names[randint(0, len(female_names))] + " " + surname
  else:
    names = [*male_names, *female_names]
    return names[randint(0, len(names))] + " " + surname
    
    
```
